# Outer product 

In this notebook, we demonstrate the outer product between two vectors for real and complex operations.

## Outer (real)

Let $\mathbf{x} \in \mathcal{R}^N$ and  $\mathbf{y} \in \mathcal{R}^M$ vectors given by: 

$$ \mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
\vdots \\
x_{N-1}
\end{bmatrix}
$$
and
$$ \mathbf{y} = \begin{bmatrix}
y_{0} \\
y_{1} \\
\vdots \\
y_{M-1}
\end{bmatrix}
$$

The outer product is defined as: 

$$ \mathbf{A} = \mathbf{x} \cdot \mathbf{y}^\top =  \mathbf{x} \otimes \mathbf{y}.$$

From this operation, we obtain a matrix with dimension $N \times M$ equal to: 

$$
\mathbf{A} = \begin{bmatrix}
x_0 y_0 & x_0 y_1 &  \cdots & x_0 y_{M-1}\\
\vdots & \vdots &  \ddots &  \vdots   \\
x_{N-1} y_0 & x_{N-1} y_1 & \cdots &  x_{N-1} y_{M-1}
\end{bmatrix}.
$$


**Pseudocode.1:**
    
    for i = 0:N-1
        for j = 0:M-1
            a[i,j] = x[i]*y[j]

**Pseudocode.2:**
    
    for i = 0:N-1
            a[i,:] = x[i]*y[:]

**Pseudocode.3:**
    
    for j = 0:M-1
        a[:,j] = x[:]*y[j]

## Outer (complex)

Let $\mathbf{x} \in \mathcal{C}^N$ e  $\mathbf{y} \in \mathcal{C}^M$ vectors with complex entries. The outer product results: 

$$ \mathbf{A} = \mathbf{A}_R + \mathcal{i}*\mathbf{A}_I $$

in which

$$\mathbf{A}_R = \mathbf{x}_R \otimes \mathbf{y}_R - \mathbf{x}_I \otimes \mathbf{y}_I  $$

e

$$ \mathbf{A}_R = \mathbf{x}_R \otimes \mathbf{y}_I + \mathbf{x}_I \otimes \mathbf{y}_R . $$

**Pseudocode:**

    A_R = outer(Real(x), Real(y)) - outer(Imag(x), Imag(y))
    A_I = outer(Real(x), Imag(y)) + outer(Imag(x), Real(y))

    A = A_R + imag*A_I
