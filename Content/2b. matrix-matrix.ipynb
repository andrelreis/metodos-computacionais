# Matrix-matrix product

This notebook shows some algorithms to perform a matrix-matrix product with real and complex entries. 

## Product for two matrices with real entries

Let $\mathbf{A \in \mathbb{R}^{N \times L}}$ and $\mathbf{B \in \mathbb{R}^{L \times M}}$ be matrices given by: 

$$
\mathbf{A} = \begin{bmatrix}
a_{00}    & \cdots & a_{0(L-1)}\\
\vdots    &        &  \vdots   \\
a_{(N-1)0}& \cdots &  a_{(N-1)(L-1)}
\end{bmatrix}
$$

and 

$$
\mathbf{B} = \begin{bmatrix}
b_{00}    & \cdots & b_{0(M-1)}\\
\vdots    &        &  \vdots   \\
b_{(L-1)0}& \cdots &  b_{(L-1)(M-1)}
\end{bmatrix}.
$$

A matrix-matrix product for these matrices is defined as 

$$ \mathbf{C} = \mathbf{A}\mathbf{B}, $$
with elements $c_{ij} = \sum_{k=0}^{L-1} a_{ik} b_{kj}$ given by

$$
\mathbf{C} = \begin{bmatrix}
\left(a_{00}b_{00} + \cdots + a_{0(L-1)}b_{(L-1)0}\right) & \cdots & \left(a_{00}b_{0(M-1)} + \cdots + a_{0(L-1)}b_{(L-1)(M-1)}\right) \\
\vdots    &        &  \vdots   \\
\left(a_{(N-1)0}b_{00} + \cdots + a_{(N-1)(L-1)}b_{(L-1)0}\right) & \cdots & \left(a_{(N-1)0}b_{0(M-1)} + \cdots + a_{(N-1)(L-1)}b_{(L-1)(M-1)}\right)
\end{bmatrix}.
$$
Notice that the result of a matrix-matrix product is a matrix $\mathbf{C}$ with dimension of $N \times M$. 

**Pseudocode.1 (triply nested) :**
    
    for i = 0:N-1
        for j = 0:M-1
            for k = 0:L-1
                C[i,j] += A[i,k]*B[k,j]

**Pseudocode.2 (dot product formulation):**
    
    for i = 0:N-1
        for j = 0:M-1
            C[i,j] = dot(A[i,:], B[:,j])

**Pseudocode.3 (row update):**
    
    for i = 0:N-1
        C[i,:] = B[:,:].T * A[i,:]

**Pseudocode.4 (column update):**
    
    for j = 0:M-1
        C[:,j] = A[:,:]*B[:,j]

**Pseudocode.5 (outer formulation):**
    
    for k = 0:L-1
        C[:,:] += outer(A[:,k], B[k,:])

## Product of two matrix with complex entries

Let $\mathbf{A \in \mathbb{C}^{N \times L}}$ be a matrix with complex entries given by:

$$ \mathbf{A} = \mathbf{A}_R + i*\mathbf{A}_I$$
and a matrix $\mathbf{B} \in \mathbb{C}^{L \times M}$ also with complex entries equal to

$$\mathbf{B} = \mathbf{B}_R + i*\mathbf{B}_I $$.

The matrix-matrix product results in a matrix $\mathbf{C}$ with dimension of $N \times M$ defined as

$$ \mathbf{C} = \mathbf{C}_R + i*\mathbf{C}_I $$
where

$$\mathbf{C}_R = \mathbf{A}_R \mathbf{B}_R - \mathbf{A}_I \mathbf{B}_I $$
and

$$ \mathbf{C}_I = \mathbf{A}_R \mathbf{B}_I + \mathbf{A}_I \mathbf{B}_R .$$

**Pseudocode:**

    C_R = matmat_real(Real(A), Real(B)) - matmat_real(Imag(A), Imag(B))
    C_I = matmat_real(Real(A), Imag(B)) + matmat_real(Imag(A), Real(B))

    C = C_R + imag*C_I
