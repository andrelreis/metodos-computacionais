# Scalar-vector product

This notebook shows a product between a scalar and a vector with real and complex operations. 

## Scalar-vector product (Real)

Let $a \in \mathcal{R}$ a scalar and $\mathbf{x} \in \mathcal{R}^{N}$ a vector with $N$ elements. The product between them is given by: 

$$ \mathbf{y} = a \mathbf{x}$$.

in which,

$$ \mathbf{y} = \begin{bmatrix}
a x_{0} \\
a x_{1} \\
\vdots \\
ax_{N-1}
\end{bmatrix} \quad .
$$

**Pseudocode**: 

    y = zeros(N)
    for i = 0:N-1
        y[i] = a*x[i]

```python
import numpy as np
```

```python
x = np.array([10.,14.,50,-90.,5.,-45.])
```

```python
a = 2.
```

```python
y_python = a*x[:]
```

```python
y_python
```

```python
y_andre = np.empty(x.size)
for i in range(x.size):
    y_andre[i] = a*x[i]
```

```python
y_andre
```

```python
np.allclose(y_python,y_andre)
```

## Scalar-vector product (complex)

Let $a \in \mathbb{C}$ and $\mathbf{x} \in \mathbb{C}^{N}$, with

\begin{equation}
a = a_R + i a_I,
\end{equation}

and

\begin{equation}
\mathbf{x} = \mathbf{x}_R + i\mathbf{x}_I.
\end{equation}

Then, the product $\mathbf{y}=a\mathbf{x}$ can be decomposed into its real and imaginary parts as

\begin{equation}
\mathbf{y} = \mathbf{y}_R + i\mathbf{y}_I,
\end{equation}

where

\begin{align}
\mathbf{y}_R &= a_R \mathbf{x}_R - a_I \mathbf{x}_I, \\
\mathbf{y}_I &= a_R \mathbf{x}_I + a_I \mathbf{x}_R.
\end{align}

**Pseudocode:**
    
    y_R = a_R*x_R[:] - a_I*x_I[:]
    y_I = a_R*x_I[:] + a_I*x_R[:]
    y = y_R + imag*y_I
