## Produto escalar-vetor (Real)

Seja $a \in \mathcal{R}$ um escalar e $\mathbf{x} \in \mathcal{R}^{N}$ um vetor com $N$ elementos. O produto entre eles é dado por: 

$$ \mathbf{y} = a \mathbf{x}$$.

em que,

$$ \mathbf{y} = \begin{bmatrix}
a x_{0} \\
a x_{1} \\
\vdots \\
ax_{N-1}
\end{bmatrix} \quad .
$$

**Pseudocódigo**: 

    y = zeros(N)
    for i = 0:N-1
        y[i] = a*x[i]

## Produto escalar-vetor (complexo)

Seja $a \in \mathcal{C}$ um escalar e $\mathbf{x} \in \mathcal{C}^{N \times 1}$ um vetor. Portanto, o vetor $\mathbf{x}$ e o escalar $a$ são dados por: 

$$ \mathbf{x} = \mathbf{x}_R + \mathcal{i}\mathbf{x}_I $$

e

$$ a = a_R + \mathcal{i} a_I$$.

Como resultado teremos: 

$$ \mathbf{y} = \mathbf{y}_R + \mathcal{i}\mathbf{y}_I $$

em que,

$$ \mathbf{y}_R = a_R*\mathbf{x}_R - a_I*x_{I}$$
e
$$ \mathbf{y}_I = a_R*\mathbf{x}_I + a_I*\mathbf{x}_R \quad .$$

**Pseudocódigo:**
    
    y_R = a_R*x_R[:] - a_I*x_I[:]
    y_I = a_R*x_I[:] + a_I*x_R[:]
    y = y_R + imag*y_I
