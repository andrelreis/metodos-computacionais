# Notation

Let be $\mathbf{x}$ a vector with dimension of $N \times 1$ given by: 



$$
\mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
\vdots \\
x_{N-1}
\end{bmatrix}_{\, N \times 1} \quad .
$$

Along the course, we use $x_{i}$ or $x[i]$ to define the $i$-th element of a vector $\mathbf{x}$. Finally, the transposition of the vector $\mathbf{x}$, denoted by the symbol "$^\top$", is a vector with dimension of $1 \times N$ represented by: 

$$
\mathbf{x}^{\top} = \begin{bmatrix}
x_{0} & x_{1} & \cdots & x_{N-1}
\end{bmatrix}_{1 \times N} \quad .
$$

## Colon notation

A handy way to represent elements of a vector or a matrix is by using the colon notation, denoted by the symbol "$[:]$". For instance, considering a vector $\mathbf{x}$ defined by:

$$
\mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
x_{2} \\
x_{3} \\
x_{4} \\
x_{5} \\
x_{6} \\
x_{7}
\end{bmatrix} \quad .
$$

Supose that we want to create a new vector by accessing **third**, **fourth**, **fifth** and **sixth** elements of the vector aforementioned, that is, $x_{2}$, $x_{3}$, $x_{4}$ e $x_{5}$. By using the colon notation, the result is:  

$$
\mathbf{x}[2:5:1] = \begin{bmatrix}
x_{2} \\
x_{3} \\
x_{4} \\
x_{5}
\end{bmatrix} \quad .
$$

For more details of slicing numpy array see [slicing numpy arrays](https://numpy.org/doc/stable/user/basics.indexing.html#basics-indexing).

```python
x = np.array([4, -5, 15.4, 23, 7, -65, 1, 0])
```

```python
x[2:5:1]
```

```python
x[2:6:1]
```

```python
x[-1:-5:-1]
```

```python
x[::-1]
```

```python
x[:]
```

## Pseudocodes

One way to represent code is through pseudocode. It allows us to express the logic of a program more clearly and accessibly, either before or after writing the actual code. Below is a description of an example that follows an approach very similar to what will be used in the next steps of the course.

### Sum of two vectors

$$
\begin{split}
\mathbf{z} &= \mathbf{x} + \mathbf{y} \\
&= \begin{bmatrix} 
    x_{0} + y_{0} \\
    x_{1} + y_{1} \\
    \vdots \\
    x_{N-1} + y_{N-1}
 \end{bmatrix}
\end{split}
$$

The **pseudocode** of this operation is:

    z = zeros(N)
    for i = 0:N-1
        z[i] = x[i] + y[i]
