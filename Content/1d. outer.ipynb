# Produto externo entre dois vetores

Neste notebook iremos mostrar um tipo de produto externo entre dois vetores, conhecido mais comumente como "outer". 

## Outer (real)

Seja $\mathbf{x} \in \mathcal{R}^N$ e  $\mathbf{y} \in \mathcal{R}^M$ vetores dados por: 

$$ \mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
\vdots \\
x_{N-1}
\end{bmatrix}
$$
e
$$ \mathbf{y} = \begin{bmatrix}
y_{0} \\
y_{1} \\
\vdots \\
y_{M-1}
\end{bmatrix}
$$

O produto externo entre eles será:

$$ \mathbf{A} = \mathbf{x} \cdot \mathbf{y}^\top =  \mathbf{x} \otimes \mathbf{y}.$$

Portanto, obtém-se uma matriz de dimensão $N \times M$ igual a: 

$$
\mathbf{A} = \begin{bmatrix}
x_0 y_0 & x_0 y_1 &  \cdots & x_0 y_{M-1}\\
\vdots & \vdots &  \ddots &  \vdots   \\
x_{N-1} y_0 & x_{N-1} y_1 & \cdots &  x_{N-1} y_{M-1}
\end{bmatrix}.
$$


**Pseudocódigo.1:**
    
    for i = 0:N-1
        for j = 0:M-1
            a[i,j] = x[i]*y[j]

**Pseudocódigo.2:**
    
    for i = 0:N-1
            a[i,:] = x[i]*y[:]

**Pseudocódigo.3:**
    
    for j = 0:M-1
        a[:,j] = x[:]*y[j]


## Outer (complexo)

Seja $\mathbf{x} \in \mathcal{C}^N$ e  $\mathbf{y} \in \mathcal{C}^M$ vetores com elementos complexos. O resultado do produto externo entre eles é dado por: 

$$ \mathbf{A} = \mathbf{A}_R + \mathcal{i}*\mathbf{A}_I $$

em que

$$\mathbf{A}_R = \mathbf{x}_R \otimes \mathbf{y}_R - \mathbf{x}_I \otimes \mathbf{y}_I  $$

e

$$ \mathbf{A}_R = \mathbf{x}_R \otimes \mathbf{y}_I + \mathbf{x}_I \otimes \mathbf{y}_R . $$

**Pseudocódigo:**

    A_R = outer(Real(x), Real(y)) - outer(Imag(x), Imag(y))
    A_I = outer(Real(x), Imag(y)) + outer(Imag(x), Real(y))

    A = A_R + imag*A_I
