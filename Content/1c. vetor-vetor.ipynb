# Produto entre dois vetores

Neste notebook iremos mostrar o produto entre dois vetores, conhecido mais comumente como produto escalar (dot). 

## Produto vetor-vetor (real)

Seja $\mathbf{x}, \mathbf{y} \in \mathcal{R}^N$ vetores com $N$ elementos dados por: 

$$ \mathbf{y} = \begin{bmatrix}
y_{0} \\
y_{1} \\
\vdots \\
y_{N-1}
\end{bmatrix}
$$
e 

$$ \mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
\vdots \\
x_{N-1}
\end{bmatrix} \quad .
$$

O produto entre eles é igual a:

$$ z = \mathbf{x}^\top \mathbf{y} = \sum_{i = 0}^{N-1} x_i y_i \quad . $$


**Pseudocódigo:**
    
    z = 0
    for i = 0:N-1
        z[i] = z[i] + x[i]*y[i]

## Produto vetor-vetor (complexo)

Seja $\mathbf{x}, \mathbf{y} \in \mathcal{C}^N$ vetores com $N$ elementos. O resultado do produto entre estes dois vetores será igual a: 

$$ z = z_R + \mathcal{i}*z_I $$

em que 

$$ z_R = \mathbf{x}_{R}^{\top} \mathbf{y}_{R} - \mathbf{x}_{I}^{\top} \mathbf{y}_I $$

e 

$$ z_I = \mathbf{x}_{R}^{\top} \mathbf{y}_I + \mathbf{x}_{I}^{\top} \mathbf{y}_R \quad .$$

**Pseudocódigo:**

    z_R = dot(Real(x), Real(y)) - dot(Imag(x), Imag(y))
    z_I = dot(Real(x), Imag(y)) + dot(Imag(x), Real(y))

    z = z_R + imag*z_I
