# Notações

Seja $\mathbf{x}$ um vetor de dimensão $N \times 1$ dado por: 



$$
\mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
\vdots \\
x_{N-1}
\end{bmatrix}_{\, N \times 1} \quad .
$$

Ao longo do curso iremos utilizar $x_{i}$ ou $x[i]$ para definir o $i$-ésimo elemento do vetor $\mathbf{x}$. Por fim, o transposto $\mathbf{x}^\top$ é um vetor de dimensão $1 \times N$ representado por: 

$$
\mathbf{x}^{\top} = \begin{bmatrix}
x_{0} & x_{1} & \cdots & x_{N-1}
\end{bmatrix}_{1 \times N} \quad .
$$

## Notação de colon

A notação de colon é um jeito rápido e esperto de selecionar elementos de um vetor, que é denotada por "
$[:]$". Por exemplo, considerando um vetor $\mathbf{x}$ definido por:

$$
\mathbf{x} = \begin{bmatrix}
x_{0} \\
x_{1} \\
x_{2} \\
x_{3} \\
x_{4} \\
x_{5} \\
x_{6} \\
x_{7}
\end{bmatrix} \quad .
$$

Agora supunha que queremos criar um novo vetor formado pelo **terceiro**, **quarto**, **quinto** e **sexto** elementos deste vetor, isto é, $x_{2}$, $x_{3}$, $x_{4}$ e $x_{5}$. Utilizando a notação de colon, teremos:  

$$
\mathbf{x}[2:5:1] = \begin{bmatrix}
x_{2} \\
x_{3} \\
x_{4} \\
x_{5}
\end{bmatrix} \quad .
$$

No entanto, ao utilizarmos esta notação em um numpy array, acontece uma pequena diferença. Para mais detalhes basta olhar em [slicing numpy arrays](https://numpy.org/doc/stable/user/basics.indexing.html#basics-indexing)

```python
x = np.array([4, -5, 15.4, 23, 7, -65, 1, 0])
```

```python
x[2:5:1]
```

```python
x[2:6:1]
```

```python
x[-1:-5:-1]
```

```python
x[::-1]
```

```python
x[:]
```

### Soma vetor-vetor

$$
\begin{split}
\mathbf{z} &= \mathbf{x} + \mathbf{y} \\
&= \begin{bmatrix} 
    x_{0} + y_{0} \\
    x_{1} + y_{1} \\
    \vdots \\
    x_{N-1} + y_{N-1}
 \end{bmatrix}
\end{split}
$$

O pseudocódigo desta operação será:

    z = zeros(N)
    for i = 0:N-1
        z[i] = x[i] + y[i]
